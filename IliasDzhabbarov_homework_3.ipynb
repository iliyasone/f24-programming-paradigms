## 3.1 Contexts


### Exercise 3.1.1

```racket
(require minikanren)
(require minikanren/matche)

(defrel (context-lookupo context var type)
    (fresh (var_p type_p tail head)
        (== (cons head tail) context)
        (conde
            [  
                (== (cons var_p type_p) head)
                (== var var_p)
                (== type type_p)
            ]
            [(context-lookupo tail var type)]
        )
    )
    
)
```

```racket
(define sample-context
'(
    (x . Int)
    (f . (Int -> Bool))
    (y . Int)
    (b . Bool)
    (+ . (Int -> (Int -> Int)))
))


(run* (type) (context-lookupo sample-context 'x type))
```

```racket
(run* (type) (context-lookupo sample-context 'z type))
```

```racket
(run* (var) (context-lookupo sample-context var 'Int))
```

```racket
(run* (var type) (context-lookupo sample-context var type))
```

### Exercise 3.1.2

```racket
(defrel (context-varso context vars)
    (fresh (context-var context-type context-tail context-head vars-head vars-tail)
        (conde
            [
                (== context '())
                (== vars '())
            ]

            [
                (== (cons vars-head vars-tail) vars)
                (== (cons context-head context-tail) context)
                
                (== (cons context-var context-type) context-head)
                (== vars-head context-var)
                    
                (context-varso context-tail vars-tail)
            ]
        )
    )
)
```

```racket
(run* (vars) (context-varso sample-context vars))
```

```racket
(run* (context) (context-varso context '(x y z)))
```

## 3.2 Untyped Expressions


```racket
(define succeed (== #t #t))
(define fail (== #f #t)) 

(define (membero x lst)
  (conde
    [(== lst '()) fail]
    [(fresh (a d)
        (== (cons a d) lst)
        (conde
            [(== x a)] 
            [(membero x d)]
        )
     )
    ]
  )
) 
```

```racket
(defrel (untyped-expro vars expr)
    (conde
        [(membero expr vars) (symbolo expr)] 
        [(membero expr vars) (numbero expr)] 
    )
)
```
