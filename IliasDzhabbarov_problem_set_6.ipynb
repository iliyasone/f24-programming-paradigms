```python
-- (c) Convert a trinary string represented as a list of 0s, 1s, and 2s into a (decimal) number

-- Helper function to convert trinary to decimal
trinaryToDecimalHelper :: [Integer] -> Integer -> Integer -> Integer
trinaryToDecimalHelper [] result _ = result
trinaryToDecimalHelper (0:xs) result digit = trinaryToDecimalHelper xs result (digit + 1)
trinaryToDecimalHelper (1:xs) result digit = trinaryToDecimalHelper xs (result + (1 * (3 ^ digit))) (digit + 1)
trinaryToDecimalHelper (2:xs) result digit = trinaryToDecimalHelper xs (result + (2 * (3 ^ digit))) (digit + 1)

-- Main function
trinaryToDecimal :: [Integer] -> Integer
trinaryToDecimal bits = trinaryToDecimalHelper (reverse bits) 0 0

trinaryToDecimal [1, 0, 2, 1, 0]
```

```python
-- (d) Return the last 3 Characters in a String. If the string is too short, return entire string

-- Helper function to return the last three characters of a string
lastThreeHelper :: String -> Int -> String
lastThreeHelper str len
    | len <= 3  = str
    | otherwise = drop (len - 3) str

-- Main function that calls the helper
lastThree :: String -> String
lastThree str = lastThreeHelper str (length str)

example :: IO ()
example = do
    print (lastThree "111011")  -- "011"
    print (lastThree "symbol")  -- "bol"
    print (lastThree "no")      -- "no"
    
example
```

```python
-- (e) Decrement a binary number (represented as a list of digits). Decrementing zero should produce zero

-- Helper function to decrement a binary number
helperDecrement :: [Integer] -> [Integer] -> Integer -> [Integer]
helperDecrement [] answer buff
    | buff > 0  = [0]
    | otherwise = reverse answer
helperDecrement (0:xs) answer 1 = helperDecrement xs (1 : answer) 1  
helperDecrement (1:xs) answer 1 = helperDecrement xs (0 : answer) 0  
helperDecrement (x:xs) answer _ = helperDecrement xs (x : answer) 0 

removeLeadingZeros :: [Integer] -> [Integer]
removeLeadingZeros [] = []
removeLeadingZeros (0:xs)
    | null xs   = [0]  
    | otherwise = removeLeadingZeros xs 
removeLeadingZeros xs = xs  

decrement :: [Integer] -> [Integer]
decrement bits = removeLeadingZeros $ reverse $ helperDecrement (reverse bits) [] 1

example :: IO ()
example = do
    print (decrement [1,0,1,1,0])  -- [1,0,1,0,1]
    print (decrement [1,0,0,0,0])  -- [1,1,1,1]
    print (decrement [0])          -- [0]
    
example
```

```python
data Day = Day Int deriving (Show)    -- day of a month (from 1 to 29, 30, or 31)
data Month = Month Int deriving (Show)  -- month of the year (from 1 to 12)
data Year = Year Int deriving (Show)    -- year

data Date = Date Day Month Year deriving (Show)  -- a valid date
```

```python
-- | Extract year from a date.
yearOf :: Date -> Year
yearOf (Date _ _ year) = year

-- | Extract month from a date.
monthOf :: Date -> Month
monthOf (Date _ month _) = month


-- | Extract day of the month from a date.
dayOf :: Date -> Day
dayOf (Date day _ _) = day

-- | A year is leap if it is divisible by 4 and
-- not divisible by 100 (except if it is divisible by 400).
isLeapYear :: Year -> Bool
isLeapYear (Year year)
  | year `mod` 400 == 0 = True
  | year `mod` 100 == 0 = False
  | year `mod` 4 == 0 = True
  | otherwise = False
 
exampleDate :: Date
exampleDate = Date (Day 15) (Month 8) (Year 2024)
  
isLeapYear (yearOf exampleDate)
```

```python
-- | How many days are in a given month (in a given year)?
daysInMonth :: Year -> Month -> Int
daysInMonth year (Month month)
  | month == 1  = 31  -- January
  | month == 2  = if isLeapYear year then 29 else 28  -- February
  | month == 3  = 31  -- March
  | month == 4  = 30  -- April
  | month == 5  = 31  -- May
  | month == 6  = 30  -- June
  | month == 7  = 31  -- July
  | month == 8  = 31  -- August
  | month == 9  = 30  -- September
  | month == 10 = 31  -- October
  | month == 11 = 30  -- November
  | month == 12 = 31  -- December
  
  
daysInMonth (yearOf exampleDate) (monthOf exampleDate)
```

```python
-- | Add some number of years (can be positive, negative, or zero) to a given date
-- to produce a new valid date.
addYears :: Int -> Date -> Date
addYears n (Date (Day day) (Month month) (Year year))
  | month == 2 && day == 29 && not (isLeapYear (Year newYear)) = Date (Day 28) (Month 2) (Year newYear)
  | otherwise = Date (Day day) (Month month) (Year newYear)
  where
    newYear = year + n
    
-- Example usage:
exampleDate :: Date
exampleDate = Date (Day 29) (Month 2) (Year 2020)


addYears 1 exampleDate  -- Result: Date (Day 28) (Month 2) (Year 2021)
```

```python
-- | Add some number of months (can be positive, negative, or zero) to a given date
-- to produce a new valid date.
addMonths :: Int -> Date -> Date
addMonths n (Date (Day day) (Month month) (Year year)) =
  let newMonth = month + n
      (extraYears, finalMonth) = (newMonth - 1) `divMod` 12
      newYear = year + extraYears
      correctedMonth = finalMonth + 1
      maxDays = daysInMonth (Year newYear) (Month correctedMonth)
      correctedDay = min day maxDays
  in Date (Day correctedDay) (Month correctedMonth) (Year newYear)
```

```python
-- | Add some number of days (can be positive, negative, or zero) to a given date
-- to produce a new valid date.
addDays :: Int -> Date -> Date
addDays n date@(Date (Day day) (Month month) (Year year))
  | n == 0 = date
  | n > 0 = addDays (n - 1) (nextDay date)
  | n < 0 = addDays (n + 1) (prevDay date)

-- Helper function to get the next day
nextDay :: Date -> Date
nextDay (Date (Day day) (Month month) (Year year))
  | day < daysInMonth (Year year) (Month month) = Date (Day (day + 1)) (Month month) (Year year)
  | month < 12 = Date (Day 1) (Month (month + 1)) (Year year)
  | otherwise = Date (Day 1) (Month 1) (Year (year + 1))

-- Helper function to get the previous day
prevDay :: Date -> Date
prevDay (Date (Day day) (Month month) (Year year))
  | day > 1 = Date (Day (day - 1)) (Month month) (Year year)
  | month > 1 = Date (Day (daysInMonth (Year year) (Month (month - 1)))) (Month (month - 1)) (Year year)
  | otherwise = Date (Day 31) (Month 12) (Year (year - 1))
```

```python
-- Example Date: February 29th, 2020 (a leap year)
exampleDate1 :: Date
exampleDate1 = Date (Day 29) (Month 2) (Year 2020)

-- Add 12 months to February 29th, 2020 (should end up in February 2021, which is not a leap year)
exampleAddMonths1 :: Date
exampleAddMonths1 = addMonths 12 exampleDate1  -- Result: Date (Day 28) (Month 2) (Year 2021)
exampleAddMonths1

-- Add -3 months to February 29th, 2020 (should end up in November 2019)
exampleAddMonths2 :: Date
exampleAddMonths2 = addMonths (-3) exampleDate1  -- Result: Date (Day 29) (Month 11) (Year 2019)
exampleAddMonths2

-- Add 1 month to January 31st, 2021 (should end up in February 28th, 2021)
exampleDate2 :: Date
exampleDate2 = Date (Day 31) (Month 1) (Year 2021)
exampleAddMonths3 :: Date
exampleAddMonths3 = addMonths 1 exampleDate2  -- Result: Date (Day 28) (Month 2) (Year 2021)
exampleAddMonths3
```

```python
-- Example Date: February 28th, 2021 (a non-leap year)
exampleDate3 :: Date
exampleDate3 = Date (Day 28) (Month 2) (Year 2021)
exampleDate3

-- Add 1 day to February 28th, 2021 (should result in March 1st, 2021)
exampleAddDays1 :: Date
exampleAddDays1 = addDays 1 exampleDate3  -- Result: Date (Day 1) (Month 3) (Year 2021)
exampleAddDays1

-- Add -5 days to March 1st, 2021 (should result in February 24th, 2021)
exampleAddDays2 :: Date
exampleAddDays2 = addDays (-5) (Date (Day 1) (Month 3) (Year 2021))  -- Result: Date (Day 24) (Month 2) (Year 2021)
exampleAddDays2

-- Add 365 days to March 1st, 2021 (should result in March 1st, 2022)
exampleAddDays3 :: Date
exampleAddDays3 = addDays 365 exampleDate3  -- Result: Date (Day 1) (Month 3) (Year 2022)
exampleAddDays3

-- Add -365 days to March 1st, 2022 (should result in March 1st, 2021)
exampleAddDays4 :: Date
exampleAddDays4 = addDays (-365) (Date (Day 1) (Month 3) (Year 2022))  -- Result: Date (Day 1) (Month 3) (Year 2021)
exampleAddDays4
```

Let's proof by Equational Reasoning


### (a)
For any date `date`, if `isLeapYear (yearOf date) = True`, then `isLeapYear (yearOf (addYears 3 date)) = False`.

1. Let `date = Date d m y`, where `y` is a leap year.
2. Since `y` is a leap year, it is divisible by 4 and either not divisible by 100, or divisible by 400.
3. We add 3 years to `y`: `addYears 3 date = Date d m (y + 3)`.
4. To check if `y + 3` is a leap year:
   - For a typical leap year, `y` is divisible by 4, and `y + 3` is not divisible by 4, meaning `y + 3` is not a leap year.
5. Therefore, `isLeapYear (yearOf (addYears 3 date)) = False`.

### (b)
For any date `date` and any integers `x` and `y`, we have `addDays (x + y) date = addDays x (addDays y date)`.

1. By the definition of `addDays`, adding `x + y` days to `date` results in:
   - `addDays (x + y) date`.
2. We can rewrite this as a sequential addition:
   - `addDays (x + y) date = addDays x (addDays y date)`.
3. This follows from the property of addition being associative in the definition of `addDays` :)
