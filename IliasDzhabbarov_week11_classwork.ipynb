```racket
(defrel (context-lookupo context var type)
    (fresh (head tail var_p type_p)
        (== (cons head tail) context) ;; context is a cons of head and tail
        (conde
            [
                (== head (cons var_p type_p)) ;; head is a pair (var_p . type_p)
                (== var var_p)
                (== type type_p)
            ]
            [(context-lookupo tail var type)]))) ;; Recursive call on tail
```

```racket
(require minikanren)
(require minikanren/matche)

(defrel (binaryo digits)
    (conde
        [(fresh (head tail)
            (== digits (cons head tail))
            (conde
                [(== head 1)]
                [(== head 0)]
            )
            (binaryo tail)
        )
        ]
        [(fresh (x)
            (== digits (list x))
            (conde
                [(== x 1)]
                [(== x 0)]
            )
        )
        ]  
    )
)

; (run* (x) (binaryo `(1 ,x 0)))
(run* (q)
 (fresh (d1 d2)
 (== q `(,d1 ,d2))
 (binaryo q)))
```

```racket
(defrel (unaryo n)
  (conde
   [(== 'z n)]
   [(fresh (m)
           (== (list 's m) n)
           (unaryo m))]))

(defrel (testo n sn)
  (conde
   [(== (list 's n) sn)
    (unaryo n)]
   [(fresh (k)
           (== (list 's k) n)
           (testo k sn))]))

; How many answers will the following query provide?
(run* (q) (testo '(s (s (s z))) q))
```
