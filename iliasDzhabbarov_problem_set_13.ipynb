Task 1

```racket
(require minikanren)
(require minikanren/matche)

(defrel (treeo tree)
    (conde
        [(== tree 'empty)]
        [(fresh (value left right)
            (== tree `(node ,value ,left ,right))
            (treeo left)
            (treeo right)
        )
        ]
    )
)


```

```racket
(run* (q) (treeo 'empty))
```

```racket
(run* (q) (treeo '(node 1 empty (node 2 (node 3 empty empty) empty))))
```

```racket
(run 5 (q) (treeo q))
```

```racket
(defrel (tree-equalo a b)
    (conde
        [(== a 'empty)
         (== b 'empty)
        ]
        
    )
)
```

```racket
(require minikanren)
(require minikanren/matche)


(defrel (prefix-subtreeo small large)
    (fresh (value left right)
        (conde
            [(== large `(node ,value ,left ,right))]
            [(== large 'empty)]
        )
        
        (let [(sl `(,small . ,large))]
            (matche sl 
                [('empty . ,large)]
                [
                `((node ,value empty empty) . ,large)
                
                ]
                
            )
        )
    )
)
```

```racket
(run* (q)
  (prefix-subtreeo q 'empty))
```

```racket
(run* (q)
    (prefix-subtreeo q
        '(node 3
            (node 1 empty empty)
            (node 4 empty empty)
        )
    )
)
```

```racket
(run* (q)
(prefix-subtreeo q
'(node 3
(node 1 empty
(node 2 empty empty))
(node 4 empty empty))))
```
